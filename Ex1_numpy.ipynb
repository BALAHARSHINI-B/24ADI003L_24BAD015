```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt

# Load the ecommerce dataset using full path
ecom_df = pd.read_csv(
    r"C:\Users\STUDENT\Downloads\archive (3)\data.csv",
    encoding="ISO-8859-1"
)

# Display first few rows
print(ecom_df.head())

# Display last few rows
print(ecom_df.tail())

# Dataset information
print(ecom_df.info())

# Statistical summary
print(ecom_df.describe())

# Check for missing values
print(ecom_df.isnull().sum())

# Sales per product (top 10)
sales = ecom_df.groupby("Description")["Quantity"].sum().head(10)

# Bar chart
sales.plot(kind="bar", figsize=(10,5))
plt.title("Top 10 Products by Quantity Sold")
plt.xlabel("Product")
plt.ylabel("Quantity")
plt.show()

# Line chart
sales.plot(kind="line", marker="o")
plt.title("Sales Trend of Top Products")
plt.xlabel("Product")
plt.ylabel("Quantity")
plt.show()
```
